# LedgerMind AI — Exploratory Data Analysis (EDA)

**Objective**: Profile credit card transaction distributions, quantify class imbalance, examine monetary amounts, and evaluate feature correlation with fraud.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

from ml.preprocessing.eda import profile_dataset
from ml.preprocessing.validator import load_dataset, validate_dataframe

# Check for full or sample dataset
data_path = Path('../data/raw/creditcard.csv')
if not data_path.is_file():
    data_path = Path('../data/raw/creditcard_sample.csv')

df = load_dataset(data_path)
summary = validate_dataframe(df)
summary.print_report()
```

### Statistical Profile & Imbalance Analysis

```python
profile = profile_dataset(df)
profile.print_summary()
```

### Transaction Amount Distribution (Logarithmic Scale)
Financial transaction amounts are heavily right-skewed. We inspect `log(1 + Amount)` to normalize scale.

```python
plt.figure(figsize=(10, 5))
sns.kdeplot(np.log1p(df[df['Class'] == 0]['Amount']), label='Legitimate', color='#2ecc71', fill=True, alpha=0.4)
sns.kdeplot(np.log1p(df[df['Class'] == 1]['Amount']), label='Fraudulent', color='#e74c3c', fill=True, alpha=0.4)
plt.title('Log Transaction Amount Distribution')
plt.xlabel('log(1 + Amount)')
plt.ylabel('Density')
plt.legend()
plt.show()
```
